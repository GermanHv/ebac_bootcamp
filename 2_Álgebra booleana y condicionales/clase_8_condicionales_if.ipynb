```python
x = 5
if x == 5: print("x es igual a 5")
if x > 4: print("x es mayor que 4")
if x >= 5: print("x es mayor o igual a 5")
if x < 6: print("x es menor que 6")
if x <= 5: print("x es menor o igual a 5")
if x != 6: print("x no es igual a 6")
```

```python
x = input("Ingrese un número: ") # esto es una cadena de texto
# Validación de negativos
if float(x) < 0: 
    print("El número es negativo")
    if float(x) <= -1:
        print("El número es menor que -1")
    if float(x) <= -10:
        print("El número es menor que -10")

if float(x) == 0: print("El número es cero")

if float(x) > 0:
    print("El número es positivo")
    if float(x) > 1:
        print("El número es mayor que 1")
        if float(x) < 100:
            print("El número es menor que 100")

        if float(x) > 100:
            print("El número es mayor que 100")

print ("fin del programa")
```
