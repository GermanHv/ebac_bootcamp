```python
x = float(input("Ingrese un número: "))
if x < 0:
    print("x = ", x)
    print("Número negativo")
elif x == 0:
    print("x = ", x)
    print("Número cero")
else:
    print("x = ", x)
    print("Número positivo")

print("\nFin del programa")
```

```python
x = float(input("Ingrese un número: "))
if x < 10:
    print("Número menor que 10")
elif x < 40:
    print("Número menor que 40")
elif x < 30:
    print("Número menor que 30")
else:
    print("otro número")
```
