```python
cadena = input("Ingrese un número entero positivo: ")
try:
    numero = int(cadena)
    if numero < 0:
        raise ValueError("El número debe ser positivo.") 
    print("El número ingresado es:", numero)
except ValueError as e:
    print("Error:", e)

if numero > 0:
    print("El número es positivo.")
else:
    print("El número es cero o negativo.")
```

### Usos avanzados de if

```python
lluvia = input("¿Está lloviendo? (sí/no): ")
try:
    if lluvia.lower() not in ["sí", "si", "no"]:
        raise ValueError("Debe responder 'sí' o 'no'.")
    if lluvia.lower() == "sí" or lluvia.lower() == "si":
        print("Recuerda llevar un paraguas.")
    else:
        print("Disfruta del buen clima.")
except ValueError as e:
    print("Error:", e)
```

```python
# Ejercicio
try:

    # Intenta realizar la división

    a = 'asd'  # Cambiar el valor del numerador para probar diferentes casos

    b = 2   # Cambiar el valor del denominador para probar diferentes casos

    resultado = a / b

except ZeroDivisionError:

    # Captura la excepción si b es cero

    print("Error: No se permite la división por cero.")

    resultado = None # None indica que no hay un resultado válido

except TypeError:

    # Captura la excepción si los tipos de los argumentos no son numéricos

    print("Error: Ambos argumentos deben ser números.")

    resultado = None

else:

    # Si no ocurre una excepción, verifica si el resultado es positivo o negativo

    if resultado > 0:

        print("El resultado es positivo.")

    elif resultado < 0:

        print("El resultado es negativo.")

    else:

        print("El resultado es cero.")

    print("El resultado de la división es:", resultado)

finally:

    # Esta cláusula se ejecuta siempre, independientemente de excepciones

    print("Ejecución del bloque try/except finalizada.")

```

En este ejemplo:

- El bloque try intenta realizar la división entre a y b.

- Si b = 0, se captura la excepción ZeroDivisionError y se imprime un mensaje indicando que no se permite dividir entre cero.

- Si b no es un número, se captura la excepción TypeError y se indica que ambos valores deben ser numéricos.

- Si no ocurre ninguna excepción, el bloque else evalúa si el resultado es positivo, negativo o cero.

- Finalmente, el bloque finally siempre se ejecuta, indicando que el proceso ha finalizado.

```python
# Ejemplo de uso de operadores lógicos y estructuras condicionales
def clasificar_cliente(saldo, transacciones):
    if saldo > 10000 and transacciones > 50:
        return "Cliente VIP"
    elif saldo > 5000 or transacciones > 20:
        return "Cliente Preferente"
    else:
        return "Cliente Regular"

# Uso de la función
cliente_tipo = clasificar_cliente(3000, 19)
print(cliente_tipo)
```
