```python
import numpy as np

a = np.array([True, False, True])

b = np.array([False, False, True])

# Uso de & y | para operaciones lógicas en arrays booleanos

result_and = a & b # Resultado: array([False, False, True])

result_or = a | b # Resultado: array([ True, False, True])

print(result_and)

print(result_or)
```
