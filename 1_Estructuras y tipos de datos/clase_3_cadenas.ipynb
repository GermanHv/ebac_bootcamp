## Cadenas

```python
nombre = "Germán"
apellido = "Horta"
nombre_completo = nombre + " " + apellido
print(nombre_completo)
```

```python
print(nombre_completo[0])  # G
print(nombre_completo[1])  # e
print(nombre_completo[2])  # r
```

```python
print(nombre_completo[0:7])  # Germán
print(nombre_completo[7:])   # Horta
print(nombre_completo[:7])   # Germán
```

```python
print(len(nombre_completo))  # 12
print(nombre_completo.upper())  # GERMÁN HORTA
print(nombre_completo.lower())  # germán horta
print(nombre_completo.split())  # ['Germán', 'Horta']
print(nombre_completo.replace("Germán", "Dayan"))  # Dayan Horta
```
