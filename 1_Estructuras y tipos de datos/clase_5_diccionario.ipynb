```python
conteo = dict()
nombres = ['José', 'Claudia', 'Ramiro', 'José', 'Claudia']
# Para cada nombre en la lista de nombres
# verificar si el nombre ya está en el diccionario de conteo
# Si el nombre no está en el diccionario, agregarlo con un conteo de 1
# Si el nombre ya está en el diccionario, incrementar su conteo en 1
for nombre in nombres:
    if nombre not in conteo:
        conteo[nombre] = 1
    else:
        conteo[nombre] += 1
conteo
```

```python
archivo = open('Trabalenguas.txt')
archivo
```

```python
conteo = dict()
for text in palabras:
    if text in conteo:
        conteo[text] += 1
    else:
        conteo[text] = 1
conteo
```

```python
# Práctica
inventario = {
    'manzanas': 10,
    'naranjas': 5,
    'peras': 8
}
inventario 
```

```python
# Modificar valores
inventario['manzanas'] = 15
inventario['naranjas'] = 10
inventario
```

```python
# Agregar un nuevo producto
inventario['uvas'] = 12
inventario
```

```python
# Eliminar un producto
inventario.pop('peras')
inventario
```

```python
# Convertir a listas
claves = list(inventario.keys())
valores = list(inventario.values())
print("inventario:")
print("Productos:", claves)
print("Cantidad:", valores)
```

```python
# Ejemplo de conteo de palabras en un texto usando diccionarios
def contar_palabras(texto):
    palabras = texto.split()
    conteo = {}
    for palabra in palabras:
        if palabra in conteo:
            conteo[palabra] += 1
        else:
            conteo[palabra] = 1
    return conteo

texto = "hola mundo hola python"
resultado = contar_palabras(texto)
print(resultado)
```
