## Tuplas
Colecciones inmutables que pueden ser relacionados con diccionarios
Secuencia que trabaja de manera similar a una lista pero la cual no se puede alterar su contenido.
Usamos paréntesis para definirla.
Son "inmutables"
Son mas eficientes que las listas.

```python
x = ('José','Alberto','Manuel')
x
```

```python
print(x[0])  # Acceder al primer elemento
```

```python
y = (5, 10, 15, 20)
print(max(y))
```

```python
z = 'Hola, mundo!'
z[2] = 'l'  # Esto generará un error porque las tuplas son inmutables
```

```python
y[0] = 100  # Esto también generará un error por la misma razón
```

```python
y.sort()  # Esto generará un error porque las tuplas no tienen el método sort()
```

```python
y.append(25)  # Esto también generará un error porque las tuplas no tienen el método append()
```

```python
dir(y)
```

```python
(x,y) = (4, 'Hola')
print(x)
print(y)
```

```python
# Convertir diccionario a tupla
diccionario = dict()
diccionario['Alberto'] = 30
diccionario['Manuel'] = 25
diccionario
```

```python
tuplas = diccionario.items()
tuplas
```

```python
d = {'d': 1, 'e': 2, 'f': 3}
tuplas = sorted(d.items())
tuplas
```

```python
# Lista de compras
compras = ['manzanas', 'bananas', 'naranjas', 'uvas']
compras.append('peras')
compras.append('kiwis')

print("Lista de compras:", compras)
```

```python
# Eliminar un elemento de la lista
compras.remove('bananas')

compras.sort()  # Ordenar la lista de compras

print("Lista de compras ordenada:", compras)
```

```python
primeros = compras[:3]  # Obtener los primeros tres elementos
print("Primeros tres elementos:", primeros)

ultimos = compras[-3:]  # Obtener los últimos tres elementos
print("Últimos tres elementos:", ultimos)
```

```python
# Tuplas
ciudad1 = ("Ciudad de México", 19.4326, -99.1332)

ciudad2 = ("Guadalajara", 20.6597, -103.3496)

ciudad3 = ("Monterrey", 25.6866, -100.3161)
```

```python
# Suma de coordenadas de dos ciudades

suma_latitudes = ciudad1[1] + ciudad2[1]

suma_longitudes = ciudad1[2] + ciudad2[2]

# Crear una nueva tupla con las sumas

suma_coordenadas = (suma_latitudes, suma_longitudes)
```

```python
# Mostrar resultados

print("Coordenadas de las ciudades:")

print("Ciudad 1:", ciudad1)

print("Ciudad 2:", ciudad2)

print("Ciudad 3:", ciudad3)

print("Suma de coordenadas entre Ciudad 1 y Ciudad 2:", suma_coordenadas)
```

```python
matriz = [[1, 2], [3, 4], [5, 6]]
print(matriz[0][1]) # 2

# Comprensión de listas
cuadrados = [x**2 for x in range(1, 6)]
print(cuadrados) # [1, 4, 9, 16, 25]
```
